```python
import os
import shutil
import random
from pathlib import Path

def distribute_data(main_data_path, output_base_path, num_clients=4):
    # Create client directories
    client_paths = []
    for i in range(num_clients):
        client_path = os.path.join(output_base_path, f'client_{i+1}')
        client_paths.append(client_path)
        
        # Create class directories for each client
        for class_name in os.listdir(main_data_path):
            os.makedirs(os.path.join(client_path, class_name), exist_ok=True)
    
    # Distribute data for each class
    for class_name in os.listdir(main_data_path):
        class_path = os.path.join(main_data_path, class_name)
        files = os.listdir(class_path)
        random.shuffle(files)
        
        # Calculate files per client
        files_per_client = len(files) // num_clients
        
        # Distribute files to clients
        for i, client_path in enumerate(client_paths):
            start_idx = i * files_per_client
            end_idx = start_idx + files_per_client if i < num_clients-1 else len(files)
            
            client_class_path = os.path.join(client_path, class_name)
            for file_name in files[start_idx:end_idx]:
                src = os.path.join(class_path, file_name)
                dst = os.path.join(client_class_path, file_name)
                shutil.copy2(src, dst)
    
    # Print distribution summary
    for client_path in client_paths:
        print(f"\nData distribution for {os.path.basename(client_path)}:")
        for class_name in os.listdir(client_path):
            class_count = len(os.listdir(os.path.join(client_path, class_name)))
            print(f"{class_name}: {class_count} images")

# Example usage
main_data_path ="C:\\Users\\sa\Desktop\\Balanced_Training1" # Your training data path
output_base_path ="C:\\Users\\sa\\Desktop\\Balanced_clients1"   # Where to save client data

distribute_data(main_data_path, output_base_path)
```
